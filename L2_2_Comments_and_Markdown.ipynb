# Explain what you are coding

It is really important to explain what your code does. Good explanations give the reader - often your future self - a clear idea of the code's purpose and how it works. There are two main ways to do this:

- **Code comments:** short notes written *inside* code cells that guide the reader through the logic, line by line.
- **Markdown cells:** text cells (like this one) that give the bigger picture, describing what a section of code is about before you get to it.

In the sections below we'll look at examples of both, along with a few best practices.

## Annotating within the code

Code comments sit inside your Python script but are ignored when the code runs. Their only job is to explain the code to human readers. Python has two main kinds of comments, which we'll look at below.

### Line comments

A line comment starts with the `#` symbol (when it is placed at the end of a line of code, PEP 8 calls it an *inline comment*). Everything after the `#` on that line is ignored by Python. These comments are usually short descriptions of a single line, or of a small group of lines. Let's look at an example.

```python
# This is a line comment. It will be ignored when this cell is run
```

When you run the cell above, nothing happens. The `#` character indicates the line contains a comment and the Python interpreter simply skips over this line.

Let's have a look at a few more examples.

```python
# This list has the names of some bus stops in Karlstad
station_names = [
    "Karlstad Drottninggatan",
    "Karlstad Karolinen",
    "Karlstad Ronndungen",
    "Karlstad Kronoparkskyrkan",
    "Karlstad Universitetet"
]
```

```python
print(station_names[-1])  # This prints the last value in the list station_names
```

```python
# This doesn't work, so I'm commenting it out for now
# my_life.append(lots_of_money)
```

In the examples above you can see some of the ways in which line comments can be used in Python. We encourage you to add line comments within your Python cells to help explain what your code does, especially if there are several lines within a given code cell.

### Multi-line text with triple quotes

PEP 8 (Python's style guide) calls several consecutive `#` lines a *block comment*. In practice, many people also use triple quotes (`'''` or `"""`) to write text over several lines or to switch off a block of code. Python turns the text into a string and, because the string is not stored anywhere, does nothing with it. (In Jupyter, a string on the last line of a cell is shown as the cell's output - you will see this below.) It makes no difference whether you use three single `'''` or three double `"""` quotation marks. Let's see some examples.

```python
""" This text will also be ignored.
Even if it is spread across multiple lines.
Cool! """
```

```python
""" The list below contains the names of some bus stops in Karlstad.
    These are only a few examples - a real dataset would contain many more. """
station_names = [
    "Karlstad Drottninggatan",
    "Karlstad Karolinen",
    "Karlstad Ronndungen",
    "Karlstad Kronoparkskyrkan",
    "Karlstad Universitetet"
]
```

```python
""" None of the code below works, commenting this out for now.
step_one = learn_to_code()
step_two = become_programmer()
step_three = ???
step_four = profit()
"""
```

> ⚠️ **Common Pitfall: `'''...'''` is really a string, not a true comment**
>
> The `#` symbol is the *only* real comment character in Python. Triple-quoted text (`''' ... '''`) is actually a **string** that Python builds and then, if you don't store it in a variable, simply ignores. That makes it convenient for switching off several lines at once, but keep two things in mind:
>
> - Placed as the very first line inside a function, a triple-quoted string becomes that function's **docstring** (its official description) - not just an ignored comment. You'll see this in the Functions lesson.
> - For everyday explanations, prefer `#` line comments. They are clearer and are what other Python programmers expect.

## Expanding with Markdown

As well as code comments, we can use Markdown cells to explain our work. Markdown is not a replacement for comments inside code cells - instead, it is a place to give the bigger picture: overall context, or the reasoning behind certain choices. Let's look at an example where a Markdown cell nicely complements the code next to it.

## Summary

Good comments and clear Markdown make your notebooks much easier to understand - especially for the person most likely to reread them: **you**, a few weeks from now. In this lesson you saw:

- **Line comments** with `#` - everything after the `#` on that line is ignored by Python. Use them for short, line-by-line explanations.
- **Multi-line strings** using triple quotes `''' ... '''` - handy for temporarily switching off several lines of code (true block comments use `#` at the start of every line).
- **Markdown cells** for the bigger picture: what a section does and *why*, rather than line-by-line detail.

A good habit is to comment the **why**, not the obvious **what**. Writing `# add 1 to counter` next to `counter = counter + 1` adds little, but `# skip the header row` explains a decision the reader could not otherwise guess.
